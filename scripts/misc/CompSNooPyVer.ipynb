```python
from astropy.io import ascii
from astropy.table import join
%matplotlib inline
import matplotlib.pylab as pl
import numpy as np
```

```python
#v1 = ascii.read("../../data/working/backup/B_ceph.csv")
#v2 = ascii.read("../../data/working/B_ceph_update3.csv")

v1 = ascii.read("../../results/forAdam/Ceph_B_update2.txt")
v2 = ascii.read("../../results/forAdam/Ceph_B_update3.txt")

t =join(v1,v2,keys='sn')
#w = np.where(t['sample_1']=='CSPII')
#t=t[w]
w1 = np.where(t['Source_1']=='CSPI')
w2 = np.where(t['Source_1']=='CSPII')

pl.plot(t['mmax_corrected_2'][w1],t['mmax_corrected_2'][w1]-t['mmax_corrected_1'][w1],'r.',label='CSPI')
pl.plot(t['mmax_corrected_2'][w2],t['mmax_corrected_2'][w2]-t['mmax_corrected_1'][w2],'b.',label='CSPII')

pl.grid()
pl.legend()
pl.xlabel('SNooPy New'), pl.ylabel('Difference (New-Old)')
pl.savefig('../../plots/comp_snoopy_mmax.pdf')
pl.show()
```

```python
pl.figure(figsize=(6,8))
pl.subplot(3,3,1)
pl.plot(t['mmax_corrected_2'],t['_2']-t['Mmax_1'],'k.',label='Mmax')
pl.grid()
pl.legend()
pl.xlabel('SNooPy New'), pl.ylabel('Difference (New-Old)')

pl.subplot(3,3,2)
pl.plot(t['st_2'],t['st_2']-t['st_1'],'k.',label='s_BV')
pl.grid()
pl.legend()

pl.subplot(3,3,3)
pl.plot(t['BV_2'],t['BV_2']-t['BV_1'],'k.',label='B-V')
pl.grid()
pl.legend()
pl.tight_layout()


pl.savefig('../../plots/comp_snoopy_csp2.pdf')
pl.show()
```
